## Código Limpio 🛠️

Transforma el código desordenado en código limpio y diseño simple.

- **Es obvio para otros programadores** ✨
    - Nombres de variables claros
    - Clases y métodos concisos
    - Evitar números mágicos
  - Todo esto hace la diferencia entre código limpio y código difícil de entender

- **No contiene duplicación** 🚫
    - No hace falta recordar hacer el mismo cambio en las copias
    - Disminuye la carga cognitiva
    - Acelera el progreso

- **Contiene un número mínimo de partes móviles** ⚡
  - Menos código = menos que mantener en la cabeza = menos mantenimiento = menos bugs
  - El código es una responsabilidad, mantenlo corto y simple

## Código Limpio 🛠️


- **Pasa todas las pruebas** ✅
  - Sabes que tu código está sucio cuando solo pasa el 95% de las pruebas
  - Sabes que estás en problemas cuando tu cobertura de pruebas es 0%

- **¡Es más fácil y económico de mantener!** 💰
  - Reduce costos a largo plazo
  - Facilita el mantenimiento
  - Mejora la eficiencia del equipo

---

*Un código limpio es una inversión en la mantenibilidad y sostenibilidad de tu proyecto.*

## Cuándo Refactorizar 🔄

> Refactorizar no es solo limpiar código, es invertir en la calidad y mantenibilidad del proyecto.

- **La Regla de Tres** 📏
  1. **Primera vez**: Solo hazlo, completa la tarea.
  2. **Segunda vez**: Aunque te moleste repetir, hazlo igual.
  3. **Tercera vez**: Es momento de refactorizar.

- **Al añadir una nueva característica** 🆕
  - Te ayuda a entender código de otros
  - Si encuentras código sucio, refactoriza primero
  - El código limpio es más fácil de comprender
  - Mejoras el código para ti y futuros desarrolladores
  - Facilita la adición de nuevas funcionalidades

## Cuándo Refactorizar 🔄

> Refactorizar no es solo limpiar código, es invertir en la calidad y mantenibilidad del proyecto.

- **Al corregir bugs** 🐛
  - Los bugs son como los insectos reales:
    - Viven en los lugares más oscuros
    - Se esconden en el código más sucio
  - Limpia tu código y los errores se revelarán por sí mismos

- **Durante la revisión de código** 👥
  - Última oportunidad de limpiar antes de hacerlo público
  - Mejores prácticas:
    - Revisar en pareja con el autor
    - Arreglar problemas simples al momento
    - Estimar tiempo para fixes más complejos

## Cómo Refactorizar 🛠️

El refactoring debe realizarse como una serie de pequeños cambios, cada uno mejorando ligeramente el código existente mientras mantiene el programa funcionando.

- **El código debe quedar más limpio** 🧹
  - Causas comunes de fallo:
    - Mezclar muchos refactorings en un solo cambio grande
    - Trabajar con código extremadamente descuidado
    - Requisitos previos:
      - Tener tests escritos
      - Reservar suficiente tiempo

- **No crear nueva funcionalidad durante el refactoring** 🚫
  - Separa los procesos:
    - Refactoring por un lado
    - Desarrollo de nuevas features por otro
  - Sepáralos en commits diferentes

- **Los tests pueden fallar por dos razones:**
    - *Error durante el refactoring*
    - *Tests demasiado específicos*:
      - Ejemplo: probar métodos privados de clases

---

*El refactoring exitoso es metódico, gradual y siempre validado por tests.*


## Code Smells - Bloaters 🎈

Son código, métodos y clases que han crecido a proporciones gigantescas que son difíciles de trabajar. Generalmente no aparecen de inmediato, sino que se acumulan con el tiempo.

- **Long Method 📜:** Un método contiene demasiadas líneas de código. Generalmente, cualquier método de más de diez líneas debería hacerte empezar a cuestionar.

- **Large Class 🐘:** Una clase contiene muchos campos, métodos o líneas de código. Las clases suelen comenzar pequeñas, pero se inflan con el tiempo.
- **Primitive Obsession 🔢:** Uso de primitivas en lugar de objetos pequeño, Uso de constantes para codificar información o Uso de strings constantes como nombres de campos
- **Long Parameter List 📝:** Más de tres o cuatro parámetros para un método. Dificulta la comprensión y el uso.
- **Data Clumps 🗃️:** Diferentes partes del código contienen grupos idénticos de variables. Estos grupos deberían convertirse en sus propias clases.

## Code Smells - Object-Orientation Abusers 🎭

Son olores de código que representan casos donde los principios de la programación orientada a objetos se aplican de manera incorrecta o incompleta. Estos patrones suelen indicar una comprensión superficial de POO.

- **Switch Statements 🔄:** Presencia de operadores `switch` complejos o secuencias extensas de declaraciones `if`. Esto suele indicar que se está violando el principio de polimorfismo y Open/Closed.

- **Temporary Field 👻:** Campos que solo obtienen valores bajo ciertas circunstancias y permanecen vacíos el resto del tiempo. Como fantasmas, estos campos solo "cobran vida" en situaciones específicas.

- **Refused Bequest 🚫:** Una subclase que utiliza solo algunos métodos y propiedades heredadas de su clase padre. Indica una jerarquía mal diseñada donde la herencia no representa una verdadera relación "es un".

- **Alternative Classes with Different Interfaces 🎭:** Clases que realizan funciones idénticas pero tienen nombres de métodos diferentes. Viola el principio DRY (Don't Repeat Yourself) y dificulta el mantenimiento.

## Code Smells - Change Preventers 🔒

Son olores de código que hacen que un cambio simple requiera múltiples modificaciones en diferentes partes del código. Como resultado, el desarrollo se vuelve más complejo y costoso, "previniendo" cambios ágiles.

- **Divergent Change 🔀:** Cuando necesitas modificar muchos métodos no relacionados al hacer cambios en una clase. Por ejemplo, al agregar un nuevo tipo de producto, tienes que cambiar los métodos de búsqueda, visualización y ordenamiento de productos.

- **Shotgun Surgery 🎯:** Cuando una modificación requiere pequeños cambios en muchas clases diferentes. Como una "perdigonada", los cambios se dispersan por todo el código, haciendo difícil no olvidar alguno.

- **Parallel Inheritance Hierarchies 🌳:** Cada vez que creas una subclase para una clase, te ves obligado a crear una subclase correspondiente para otra clase. Las jerarquías crecen en paralelo, duplicando la complejidad.

## Code Smells - Dispensables 🗑️

Son elementos prescindibles en el código cuya eliminación lo haría más limpio, eficiente y fácil de entender. Como la limpieza de primavera, identificar y eliminar estos elementos mejora la calidad del código.

- **Comments 📝:** Un método está lleno de comentarios explicativos. Si necesitas explicar tanto el código con comentarios, probablemente el código no es lo suficientemente claro por sí mismo. El código debe ser autodocumentado.

- **Duplicate Code 📋:** Fragmentos de código que son prácticamente idénticos. Viola el principio DRY (Don't Repeat Yourself) y hace que los cambios requieran modificaciones en múltiples lugares.

- **Lazy Class 😴:** Clases que no hacen lo suficiente para justificar su existencia. Como un empleado perezoso, si una clase no aporta suficiente valor, debería ser eliminada para reducir el mantenimiento.

- **Data Class 📦:** Clases que solo contienen campos y métodos crud (getters y setters). Son meros contenedores de datos utilizados por otras clases, sin funcionalidad propia ni capacidad de operar sus datos independientemente.

- **Dead Code 💀:** Código que ya no se usa (variables, parámetros, campos, métodos o clases obsoletas). Como un zombi, está muerto pero sigue ocupando espacio y complicando el mantenimiento.

- **Speculative Generality 🔮:** Elementos no utilizados (clases, métodos, campos o parámetros) que se crearon anticipando funcionalidades futuras que nunca llegaron. No especules sobre el futuro, diseña para el presente.

## Code Smells - Couplers 🔗

Son olores de código que contribuyen al acoplamiento excesivo entre clases o muestran los problemas que surgen cuando el acoplamiento se reemplaza por delegación excesiva. Crean dependencias no deseadas que hacen el código más rígido y difícil de mantener.

- **Feature Envy 😍:** Un método accede más a los datos de otro objeto que a sus propios datos. Como un vecino entrometido, está más interesado en las propiedades de otra clase que en las suyas. Viola el principio de encapsulamiento y sugiere que el método debería estar en la otra clase.

- **Inappropriate Intimacy 💑:** Una clase utiliza campos y métodos internos de otra clase. Como una relación demasiado íntima, viola la privacidad y crea un acoplamiento fuerte entre clases que deberían mantener sus límites.

- **Message Chains ⛓️:** Cadenas de llamadas en el código del tipo `$a->b()->c()->d()`. Como una cadena de personas jugando al teléfono descompuesto, crea dependencias en cascada y viola la Ley de Demeter ("solo habla con tus amigos inmediatos").

- **Middle Man 👨‍💼:** Una clase que solo realiza una acción: delegar trabajo a otra clase. Como un intermediario innecesario, no aporta valor y complica la estructura del código. Si solo delega, ¿por qué existe?


## Code Smells - Other Smells 🌪️

Son olores de código que no encajan perfectamente en otras categorías pero que igualmente representan problemas significativos en el diseño y mantenimiento del código.

- **Incomplete Library Class 📚:** Las bibliotecas dejan de satisfacer las necesidades del usuario, pero al ser de solo lectura, no pueden modificarse. Como un libro con páginas faltantes, tenemos que vivir con sus limitaciones o buscar alternativas más completas.

- **Non-Idiomatic Code 🎭:** Código que no sigue las convenciones y prácticas comunes del lenguaje. Como hablar un idioma con acento extranjero muy marcado:

  - Usar bucles tradicionales en Python cuando list comprehensions serían más apropiadas
  - No utilizar `foreach` en PHP cuando iteras sobre arrays
  - Implementar getters/setters en Python en lugar de usar propiedades
  - Usar concatenación de strings con + en Python en lugar de f-strings
  - No aprovechar las características propias del lenguaje como async/await, decoradores, o generadores

```python
class Employee:
    ENGINEER = 0
    SALESMAN = 1
    MANAGER = 2

    def __init__(self, type):
        self._type = type
        
    def get_type(self):
        return self._type
        
    def calculate_pay(self):
        if self._type == Employee.ENGINEER:
            return self._monthlySalary
        elif self._type == Employee.SALESMAN:
            return self._monthlySalary + self._commission
        elif self._type == Employee.MANAGER:
            return self._monthlySalary + self._bonus

emp = Employee(2)
```
    
</div>

**Solución**

Reemplazar el código de tipo con un objeto estado. Si es necesario reemplazar un valor de campo con código de tipo, se "conecta" otro objeto estado.

```python
class EmployeeType:
    pass

class Engineer(EmployeeType):
    def calculate_pay(self, employee):
        return employee._monthlySalary

class Salesman(EmployeeType):
    def calculate_pay(self, employee):
        return employee._monthlySalary + employee._commission

class Manager(EmployeeType):
    def calculate_pay(self, employee):
        return employee._monthlySalary + employee._bonus

# =============================================================

class Employee:
    def __init__(self, type):
        self._type = type
            
    def calculate_pay(self):
        return self._type.calculate_pay(self)

emp = Employee(Manager())
```

```python
class Vector:
    def __init__(self):
        self._container = []

    def isEmpty(self):
        return len(self._container) == 0
    
    def push(self, element):
        self._cointainer.append(element)
        
    def pop(self, idx):
        self._cointainer.pop(idx)


class Stack(Vector):
    # Solo usa isEmpty() 

    def push(self, element):
        self.append(element)
        
    def pop(self):  # siempre 0
        self.pop(-1)
    
```

**Solución**

Crear un campo y poner un objeto de la superclase en él, delegar métodos al objeto de la superclase y eliminar la herencia.

```python
class Vector:
    def __init__(self):
        self._container = []

    def isEmpty(self):
        return len(self._container) == 0
    
    def push(self, element):
        self._cointainer.append(element)
        
    def pop(self, idx):
        self._cointainer.pop(idx)


class Stack:
    def __init__(self):
        self._vector = Vector()

    def isEmpty(self):
        return self._vector.isEmpty()    

    def push(self, element):
        self._vector.push(element)
        
    def pop(self):
        self._vector.pop(-1)
```
