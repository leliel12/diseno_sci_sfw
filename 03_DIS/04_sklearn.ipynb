## Obsolescence

- La norma internacional `IEC 62402:2019` de Gestión de la Obsolescencia define la obsolescencia como:

> .. la transición de disponible a no disponible por parte del fabricante en acuerdo con la especificación original"

- La obsolescencia es precedida por un declive gradual en popularidad.
- **Gestión de la Obsolescencia:** Implica actividades para mitigar los efectos de la obsolescencia, como compras de último momento, monitoreo y otras estrategias.




## Obsolescence

- **Obsolescencia Técnica:** Sucede cuando un nuevo producto o tecnología reemplaza al anterior, como la transición de videocasetes a DVD o del telégrafo al teléfono. También puede ocurrir cuando una versión más nueva de un producto reemplaza a la anterior.

- **Obsolescencia Funcional:** Ocurre cuando un producto ya no puede cumplir adecuadamente su función original, lo que puede deberse a cambios en la tecnología complementaria o a que su adaptación a nuevas necesidades es poco práctica.

- **Obsolescencia Arquitectónica:** Se aplica al entorno construido, donde los edificios pueden volverse obsoletos debido a la competencia de nuevas estructuras o cambios en su función.

- **Obsolescencia Planeada:** Algunas veces, los fabricantes introducen la obsolescencia a propósito para estimular compras repetidas a corto plazo.

- **Obsolescencia de Inventario:** Sucede cuando las existencias de productos no se venden lo suficientemente rápido, lo que afecta negativamente el flujo de efectivo de una organización.

- **Obsolescencia de Estilo:** Cuando un producto ya no es deseable debido a cambios en la moda, como la obsolescencia de prendas de vestir por tendencias de estilo.





## Deprecation <small>Depreciación</small>

Es el *desincentivo* de usar ciertos términos, características, diseños o prácticas, por lo general porque han sido reemplazados o ya no se consideran eficientes o seguros, **sin eliminarlos por completo o prohibir su uso.**

A menudo, los componentes deprecados no se eliminan por completo para garantizar la compatibilidad con sistemas heredados o como respaldo en caso de que los nuevos métodos no funcionen en escenarios específicos.

También puede implicar que una característica, diseño o práctica se eliminará o se suspenderá por completo en el futuro.






##  Razones comunes para la Depreciación


1. Reemplazo por una alternativa más potente.
2. Flaw de diseño o seguridad en la característica.
3. Considerada innecesaria y será eliminada.
4. Cambios estructurales importantes en futuras versiones.
5. Estandarización o consistencia en la nomenclatura.
6. Combinación de características independientes.

## Los warnings

En Python (como ya sabemos) se puede desactivar desde linea de comando, o desde codigo algun tipo de `warning`


```python
import warnings
warnings.filterwarnings("ignore", category=UserWarning)
```

donde:

- `action`:  one of "error", "ignore", "always", "default", "module", or "once".
- `message`:  a regex that the warning message must match.
- `category`:  a class that the warning must be a subclass of.
- `module`:  a regex that the module name must match.
- `lineno`:  an integer line number, 0 matches all warnings.
- `append`:  if true, append to the list of filters.

## Deprecation warnings

`DeprecationWarning` y `PendingDeprecationWarning` son tipos de advertencias que se utilizan para informar a los desarrolladores sobre características o comportamientos que se consideran obsoletos y que se eliminarán en futuras versiones de Python. 

### `PendingDeprecationWarning`:

- Se utiliza para marcar una característica como obsoleta en el futuro, pero aún no se ha establecido una fecha específica para su eliminación.
- Las características marcadas con `PendingDeprecationWarning` aún son funcionales y pueden usarse en la versión actual de Python, pero se recomienda encarecidamente que los desarrolladores eviten depender de ellas y busquen alternativas.
- **Esta** silenciada por defecto.

````python
def foo():
    warnings.warn("Esto se va a deprecar en algun momento", PendingDeprecationWarning)
    
foo()
````

## Deprecation warnings

`DeprecationWarning` y `PendingDeprecationWarning` son tipos de advertencias que se utilizan para informar a los desarrolladores sobre características o comportamientos que se consideran obsoletos y que se eliminarán en futuras versiones de Python. 

### `DeprecationWarning`:

- Es una advertencia que se utiliza para indicar que una característica específica se considera obsoleta y que los desarrolladores deben dejar de usarla. Esto le da a los desarrolladores tiempo para ajustar su código antes de que la característica se elimine por completo.
- Se espera que los desarrolladores dejen de usar la característica obsoleta y en su lugar utilicen las alternativas o enfoques recomendados.
- **No** esta silenciada por defecto.


````python
def foo():
    warnings.warn("Codigo obsoleto no lo uses", DeprecationWarning)
    
foo()
````

## Stacklevels

la advertencia se refiera donde se usa exactamente el código obsoleto, conviene establecer `stacklevel=2`.

````python
def foo():
    warnings.warn("Codigo obsoleto no lo uses", DeprecationWarning, stacklevel=2)
    
def baz():
    foo()
    
baz()  # fijarse el numero de linea
````

## Deprecar una clase

- Queda evidente en el uso como se depreca una función o un método. 
- Con las clases hay que tener un poco mas de cuidado

````python
class ADeprecatedClass(object):

    def __init_subclass__(cls, **kwargs):
        """This throws a deprecation warning on subclassing."""
        cls_name = cls.__name__
        warnings.warn(f'{cls_name} will be deprecated.', DeprecationWarning, stacklevel=2)
        super().__init_subclass__(**kwargs)

    def __init__(self, *args, **kwargs):
        """This throws a deprecation warning on initialization."""
        cls_name = type(self).__name__
        warnings.warn(f'{cls_name} will be deprecated.', DeprecationWarning, stacklevel=2)
        super().__init__(*args, **kwargs)
````

````python
class Subclass(ADeprecatedClass): pass  # herencia
````

````python
Subclass()  # instanciacion
````

## Deprecated 

un decorador Python `@deprecated` para deprecar clases, funciones o métodos antiguos.

```bash
$ pip install Deprecated
```

Tambien agrega documentación automática respecto a que algo esta deprecado: https://pypi.org/project/Deprecated/


````python
from deprecated import deprecated

@deprecated(version='1.2.1', reason="You should use another function")
def some_old_function(x, y): return x + y

class SomeClass(object):
    @deprecated(version='1.3.0', reason="This method is deprecated")
    def some_old_method(self, x, y): return x + y

some_old_function(12, 34)
obj = SomeClass()
obj.some_old_method(5, 8)
````

## Finalmente: Como Deprecar

- Si ya liberaron una version y quieren mantener su base de usuarios: deprequen amablemente.
- Den alternativas a la misma operación en `reason`.
- Una vez que haya pasado suficiente tiempo y los usuarios hayan tenido la oportunidad de adaptarse, eliminar la característica obsoleta en una versión futura siguiendo el cronograma que se haya comunicado.
- Para eliminar la forma más facil es convertir el warning en una exception con:

```bash
$ python -W error::DeprecationWarning
```
